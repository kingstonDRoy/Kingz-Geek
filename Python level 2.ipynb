# Pandas - Panel Data
- Pandas is built over Numpy
- DataFrame is pretty much like a spreadsheet
- Series is a one-dimensional labelled array capable of holding data of any type(int, float, string etc.)

```python
# Import necessary packages
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
```

```python
s_data= Series([2,3,4.0,5,'6']) 
s_data
```

# Selecting and Retrieving

```python
s_data[4]    # Retrieving data as per index
```

```python
s_data= Series([2,3,4.0,5],index=['row1','row2','row3','row4'])
s_data     # Auto-typecasted into a float
```

```python
s_data['row2'] # Selecting by a label
```

```python
series_data2=Series([34,23,20])
series_data2
```

```python
series_data2[2]
```

```python
# Row label assignment with index 
series_data2.index=['row1','row2','row3']
series_data2
```

```python
# Label assignment during initiation
series1=Series([2,3,4,5],index=['row1','row2','row3','row4'])
series1
```

```python
# Retrieve multiple values with label as reference
series2=Series(np.arange(8),index=['row1','row2','row3','row4','row5','row6','row7','row8'])
series2[['row2','row5','row1']]
```

```python
series2
```

```python
# Retrieve with index through slicing
series2[0:4]
```

```python
# Retrieve records by label with slicing
series2['row2':'row4']
```
