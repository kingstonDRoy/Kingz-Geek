# Treating missing values in Python

```python
cars.shape
```

```python
cars.isnull().sum()
```

```python
cars.loc[5,'hp']=np.nan
cars.loc[3,'hp']=np.nan
cars.loc[6,'hp']=np.nan
cars.loc[5,'mpg']=np.nan
cars.loc[2,'mpg']=np.nan
cars.head(10)
```

```python
cars.isnull().sum()
```

# Filling missing values with fillna()

```python
cars.fillna(150)
```

```python
cars.head()
```

```python
# Forward filling - Filling the value above the missing value
cars.fillna(method='ffill')
```

```python
# Backward filling - Filling the value below missing value
cars.fillna(method='bfill')
```

```python
# Treating the missing value with mean
cars.fillna(cars.mean()) # normally leads better results
```

# Grouping

```python
cars_groups=cars.groupby(cars.gear)
cars_groups.mean()
```

```python
cars_am=cars.groupby(cars.am)
cars_am.mean()
```

```python
cars_groups = cars.groupby([cars.gear,cars.am])
cars_groups.mean()
```

```python
cars_groups.count()
```
